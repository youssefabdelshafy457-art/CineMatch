```python
movies_merged = movies.merge(
    credits,
    left_on="id",
    right_on="movie_id",
    how="inner"
)

print("Merged shape:", movies_merged.shape)
```

```python
import ast

def extract_names(text):
    data = ast.literal_eval(text)
    return [item["name"] for item in data]
```

```python
extract_names(movies_merged.loc[0, "genres"])
```

```python
movies_merged["genres"] = movies_merged["genres"].apply(extract_names)
movies_merged["keywords"] = movies_merged["keywords"].apply(extract_names)
```

```python
movies_merged.columns.tolist()
```

```python
movies_merged = movies_merged.rename(columns={"title_x": "title"})
movies_merged = movies_merged.drop(columns=["title_y"])
```

```python
movies_merged[["title", "genres", "keywords"]].head()
```

```python
def extract_cast(text):
    data = ast.literal_eval(text)
    return [item["name"] for item in data[:3]]
```

```python
extract_cast(movies_merged.loc[0, "cast"])
```

```python
def extract_director(text):
    data = ast.literal_eval(text)

    for item in data:
        if item["job"] == "Director":
            return item["name"]

    return ""
```

```python
extract_director(movies_merged.loc[0, "crew"])
```

```python
movies_merged["cast"] = movies_merged["cast"].apply(extract_cast)
movies_merged["director"] = movies_merged["crew"].apply(extract_director)
```

```python
movies_merged[["title", "genres", "keywords", "cast", "director"]].head()
```

```python
movies_merged["overview"] = movies_merged["overview"].fillna("")
```

```python
movies_merged["overview"].isnull().sum()
```

```python
movies_merged["genres"] = movies_merged["genres"].apply(lambda x: " ".join(x))
movies_merged["keywords"] = movies_merged["keywords"].apply(lambda x: " ".join(x))
movies_merged["cast"] = movies_merged["cast"].apply(lambda x: " ".join(x))
```

```python
movies_merged[["title", "genres", "keywords", "cast", "director"]].head()
```

```python
movies_merged["combined_features"] = (
    movies_merged["overview"] + " " +
    movies_merged["genres"] + " " +
    movies_merged["keywords"] + " " +
    movies_merged["cast"] + " " +
    movies_merged["director"]
)
```

```python
movies_merged[["title", "combined_features"]].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(
    stop_words="english"
)
```

```python
tfidf_matrix = tfidf.fit_transform(
    movies_merged["combined_features"]
)
```

```python
print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity_matrix = cosine_similarity(tfidf_matrix)
print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
movies_merged[movies_merged["title"] == "Avatar"].index
```

```python
avatar_index = movies_merged[movies_merged["title"] == "Avatar"].index[0]

similarity_scores = list(enumerate(similarity_matrix[avatar_index]))

similarity_scores = sorted(
    similarity_scores,
    key=lambda x: x[1],
    reverse=True
)

similarity_scores[1:6]
for index, score in similarity_scores[1:6]:
    print(movies_merged.iloc[index]["title"], "→", score)

```

```python
import requests

url = "https://en.wikipedia.org/w/api.php"

params = {
    "action": "query",
    "format": "json",
    "generator": "search",
    "gsrsearch": "Avatar film",
    "gsrlimit": 5,
    "prop": "pageimages",
    "piprop": "thumbnail",
    "pithumbsize": 500,
}

response = requests.get(
    url,
    params=params,
    headers={"User-Agent": "CineMatch/1.0"},
    timeout=10
)

print(response.status_code)

data = response.json()

pages = data.get("query", {}).get("pages", {})

for page in pages.values():
    print(page.get("title"))
    print(page.get("thumbnail", {}).get("source"))
    print("---")
```
